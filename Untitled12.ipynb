```python
import time
import warnings
import numpy as np
import matplotlib.pyplot as plt
import psutil
from scipy import ndimage

import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader

warnings.filterwarnings('ignore')

SEED = 42
np.random.seed(SEED)
torch.manual_seed(SEED)
if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)

DEVICE = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"Execution device: {DEVICE}")
if torch.cuda.is_available():
    print(f"GPU: {torch.cuda.get_device_name(0)}")

class UltimateWheelGenerator:
    """Procedural Wheel Generator with morphological volume tuning and symmetry controls."""

    def __init__(self):
        self.patterns = {
            'standard': self._generate_standard,
            'web': self._generate_web,
            'spiral': self._generate_spiral,
            'sport': self._generate_sport,
            'luxury': self._generate_luxury,
            'hybrid': self._generate_hybrid
        }

    def generate_wheel(self, vol_frac=0.25, num_spokes=6, size=128,
                       pattern_type='standard', complexity=0.5, symmetry=0.8):
        if pattern_type not in self.patterns:
            pattern_type = 'standard'
        return self.patterns[pattern_type](vol_frac, num_spokes, size, complexity, symmetry)

    def _generate_standard(self, vol_frac, num_spokes, size, complexity, symmetry):
        y, x = np.ogrid[-size//2:size//2, -size//2:size//2]
        r = np.sqrt(x**2 + y**2)
        theta = np.arctan2(y, x)
        img = np.zeros((size, size))

        hub_radius = size * (0.08 + 0.04 * complexity)
        img[r < hub_radius] = 1.0

        rim_inner = size * (0.32 + 0.08 * complexity)
        rim_thickness = size * (0.03 + 0.02 * complexity)
        rim_outer = rim_inner + rim_thickness
        img[(r >= rim_inner) & (r <= rim_outer)] = 1.0

        if num_spokes > 0:
            spoke_width = size * 0.012 * (1 + 0.5 * complexity)
            base_angles = np.linspace(0, 2 * np.pi, num_spokes, endpoint=False)

            # Use symmetry parameter to inject angular perturbation
            asymmetry_jitter = (1.0 - symmetry) * (np.pi / max(num_spokes, 1)) * 0.4
            jitter = np.sin(np.arange(num_spokes) * 1.5) * asymmetry_jitter
            spoke_angles = base_angles + jitter

            for angle in spoke_angles:
                angle_diff = np.abs(np.mod(theta - angle + np.pi, 2 * np.pi) - np.pi)
                radial_factor = 0.5 + 0.5 * (r - hub_radius) / max(1.0, rim_inner - hub_radius)
                current_width = spoke_width * radial_factor
                spoke_mask = (angle_diff < current_width / size * np.pi) & (r >= hub_radius) & (r <= rim_inner)
                img[spoke_mask] = 1.0

        return self._advanced_volume_adjustment(img, vol_frac)

    def _generate_web(self, vol_frac, num_spokes, size, complexity, symmetry):
        y, x = np.ogrid[-size//2:size//2, -size//2:size//2]
        r = np.sqrt(x**2 + y**2)
        theta = np.arctan2(y, x)
        img = np.zeros((size, size))

        web_radius = size * 0.38
        primary_spokes = max(4, int(num_spokes * (1.0 + 0.5 * complexity)))
        base_angles = np.linspace(0, 2 * np.pi, primary_spokes, endpoint=False)

        jitter = (1.0 - symmetry) * 0.15 * np.cos(np.arange(primary_spokes))
        for angle in base_angles + jitter:
            angle_diff = np.abs(np.mod(theta - angle + np.pi, 2 * np.pi) - np.pi)
            img[(angle_diff < 0.02 * np.pi) & (r < web_radius)] = 1.0

        ring_count = max(2, int(4 * complexity))
        for i in range(1, ring_count + 1):
            ring_r = web_radius * (i / ring_count)
            img[(r >= ring_r - 1.5) & (r <= ring_r + 1.5)] = 1.0

        return self._advanced_volume_adjustment(img, vol_frac)

    def _generate_spiral(self, vol_frac, num_spokes, size, complexity, symmetry):
        y, x = np.ogrid[-size//2:size//2, -size//2:size//2]
        r = np.sqrt(x**2 + y**2)
        theta = np.arctan2(y, x)
        img = np.zeros((size, size))

        spiral_count = max(2, int(num_spokes))
        tightness = 1.5 + 2.0 * complexity

        for idx in range(spiral_count):
            phase = 2 * np.pi * idx / spiral_count + (1.0 - symmetry) * 0.2 * (idx % 2)
            spiral = (theta + phase - tightness * np.log(r + 1)) % (2 * np.pi)
            img[(np.abs(spiral - np.pi) < 0.1) & (r < size * 0.38)] = 1.0

        return self._advanced_volume_adjustment(img, vol_frac)

    def _generate_sport(self, vol_frac, num_spokes, size, complexity, symmetry):
        return self._generate_standard(vol_frac, max(3, num_spokes), size, complexity * 0.8, symmetry)

    def _generate_luxury(self, vol_frac, num_spokes, size, complexity, symmetry):
        base = self._generate_standard(vol_frac * 0.6, num_spokes, size, complexity, symmetry)
        web = self._generate_web(vol_frac * 0.4, num_spokes * 2, size, complexity, symmetry)
        return self._advanced_volume_adjustment(np.maximum(base, web), vol_frac)

    def _generate_hybrid(self, vol_frac, num_spokes, size, complexity, symmetry):
        s = self._generate_standard(vol_frac * 0.5, num_spokes, size, complexity, symmetry)
        sp = self._generate_spiral(vol_frac * 0.5, num_spokes, size, complexity, symmetry)
        return self._advanced_volume_adjustment(np.maximum(s, sp), vol_frac)

    def _advanced_volume_adjustment(self, img, target_vf):
        current_vf = img.mean()
        tolerance = 0.005
        max_iter = 15
        iter_count = 0

        img_adj = (img > 0.5).astype(np.float32)

        while abs(current_vf - target_vf) > tolerance and iter_count < max_iter:
            if current_vf < target_vf:
                dist = ndimage.distance_transform_edt(1.0 - img_adj)
                boundary_mask = (dist > 0) & (dist <= 1.5)
                pixels_needed = int((target_vf - current_vf) * img_adj.size)
                cand_y, cand_x = np.where(boundary_mask)
                if len(cand_y) == 0:
                    break
                select_size = min(pixels_needed, len(cand_y))
                indices = np.random.choice(len(cand_y), size=select_size, replace=False)
                img_adj[cand_y[indices], cand_x[indices]] = 1.0
            else:
                dist = ndimage.distance_transform_edt(img_adj)
                boundary_mask = (dist > 0) & (dist <= 1.5)
                pixels_to_remove = int((current_vf - target_vf) * img_adj.size)
                cand_y, cand_x = np.where(boundary_mask)
                if len(cand_y) == 0:
                    break
                select_size = min(pixels_to_remove, len(cand_y))
                indices = np.random.choice(len(cand_y), size=select_size, replace=False)
                img_adj[cand_y[indices], cand_x[indices]] = 0.0

            current_vf = img_adj.mean()
            iter_count += 1

        return self._apply_final_touches(img_adj)

    def _apply_final_touches(self, img):
        img_binary = (img > 0.5).astype(np.uint8)
        labeled, num_features = ndimage.label(img_binary)
        if num_features > 1:
            sizes = ndimage.sum(img_binary, labeled, range(1, num_features + 1))
            main_label = np.argmax(sizes) + 1
            img_binary = (labeled == main_label).astype(np.float32)
        else:
            img_binary = img_binary.astype(np.float32)
        return img_binary

wheel_engine = UltimateWheelGenerator()

class RefactoredDesignAnalyzer:
    """Topological analyzer with IoU foreground symmetry metrics."""

    def _calculate_iou_symmetry(self, img_binary):
        def iou(a, b):
            intersection = np.logical_and(a, b).sum()
            union = np.logical_or(a, b).sum()
            return intersection / union if union > 0 else 1.0

        sym_h = iou(img_binary, np.fliplr(img_binary))
        sym_v = iou(img_binary, np.flipud(img_binary))
        sym_d1 = iou(img_binary, img_binary.T)
        sym_rot180 = iou(img_binary, np.rot90(img_binary, 2))

        return {
            'symmetry_h_iou': sym_h,
            'symmetry_v_iou': sym_v,
            'symmetry_d1_iou': sym_d1,
            'symmetry_rot180_iou': sym_rot180,
            'symmetry_avg_iou': (sym_h + sym_v + sym_d1 + sym_rot180) / 4.0
        }

    def comprehensive_analysis(self, img):
        img_binary = (img > 0.5).astype(np.uint8)
        if np.sum(img_binary) == 0:
            return {'volume_fraction': 0.0, 'num_components': 0, 'connectivity_ratio': 0.0}

        volume_fraction = img_binary.mean()
        labeled, num_components = ndimage.label(img_binary)
        component_sizes = np.array([np.sum(labeled == i) for i in range(1, num_components + 1)])
        connectivity_ratio = np.max(component_sizes) / np.sum(component_sizes) if num_components > 0 else 0.0

        com = ndimage.center_of_mass(img_binary)
        center_offset = np.sqrt((com[0] - img.shape[0]/2)**2 + (com[1] - img.shape[1]/2)**2) / (img.shape[0]/2)

        metrics = {
            'volume_fraction': volume_fraction,
            'num_components': num_components,
            'connectivity_ratio': connectivity_ratio,
            'center_offset': center_offset
        }
        metrics.update(self._calculate_iou_symmetry(img_binary))
        return metrics

analyzer = RefactoredDesignAnalyzer()

class SelfAttention(nn.Module):
    def __init__(self, in_dim):
        super().__init__()
        self.query = nn.Conv2d(in_dim, in_dim // 8, 1)
        self.key = nn.Conv2d(in_dim, in_dim // 8, 1)
        self.value = nn.Conv2d(in_dim, in_dim, 1)
        self.gamma = nn.Parameter(torch.zeros(1))
        self.softmax = nn.Softmax(dim=-1)

    def forward(self, x):
        B, C, W, H = x.size()
        q = self.query(x).view(B, -1, W * H).permute(0, 2, 1)
        k = self.key(x).view(B, -1, W * H)
        energy = torch.bmm(q, k)
        attn = self.softmax(energy)
        v = self.value(x).view(B, -1, W * H)
        out = torch.bmm(v, attn.permute(0, 2, 1)).view(B, C, W, H)
        return self.gamma * out + x

class ResidualBlock(nn.Module):
    def __init__(self, channels):
        super().__init__()
        self.block = nn.Sequential(
            nn.BatchNorm2d(channels),
            nn.LeakyReLU(0.2, True),
            nn.Conv2d(channels, channels, 3, 1, 1, bias=False),
            nn.BatchNorm2d(channels),
            nn.LeakyReLU(0.2, True),
            nn.Conv2d(channels, channels, 3, 1, 1, bias=False),
        )
    def forward(self, x):
        return x + self.block(x)

class RefactoredGenerator(nn.Module):
    def __init__(self, z_dim=256, condition_dim=4, img_size=128, base_filters=64):
        super().__init__()
        self.z_dim = z_dim
        self.init_size = img_size // 16

        self.condition_net = nn.Sequential(
            nn.Linear(condition_dim, z_dim // 2),
            nn.LayerNorm(z_dim // 2),
            nn.LeakyReLU(0.2, True),
            nn.Linear(z_dim // 2, z_dim),
            nn.LayerNorm(z_dim),
            nn.LeakyReLU(0.2, True)
        )

        self.fc = nn.Sequential(
            nn.Linear(z_dim * 2, base_filters * 16 * self.init_size * self.init_size),
            nn.LayerNorm(base_filters * 16 * self.init_size * self.init_size),
            nn.ReLU(True)
        )

        self.upsample_blocks = nn.ModuleList([
            self._upsample_block(base_filters * 16, base_filters * 8),
            self._upsample_block(base_filters * 8, base_filters * 4),
            self._upsample_block(base_filters * 4, base_filters * 2),
            self._upsample_block(base_filters * 2, base_filters),
        ])
        self.attention = SelfAttention(base_filters * 2)
        self.final_conv = nn.Sequential(
            nn.Conv2d(base_filters, base_filters // 2, 3, 1, 1),
            nn.BatchNorm2d(base_filters // 2),
            nn.LeakyReLU(0.2, True),
            nn.Conv2d(base_filters // 2, 1, 3, 1, 1),
            nn.Sigmoid()
        )

    def _upsample_block(self, in_c, out_c):
        return nn.Sequential(
            nn.Upsample(scale_factor=2, mode='nearest'),
            nn.Conv2d(in_c, out_c, 3, 1, 1, bias=False),
            nn.BatchNorm2d(out_c),
            nn.LeakyReLU(0.2, True),
            ResidualBlock(out_c)
        )

    def forward(self, z, cond):
        c_proj = self.condition_net(cond)
        x = torch.cat([z, c_proj], dim=1)
        x = self.fc(x).view(-1, 1024, self.init_size, self.init_size)
        for i, block in enumerate(self.upsample_blocks):
            x = block(x)
            if x.size(-1) == 64:
                x = self.attention(x)
        return self.final_conv(x)

class RefactoredDiscriminator(nn.Module):
    """Spectral Normalization Discriminator for stable WGAN-GP learning."""
    def __init__(self, img_size=128, condition_dim=4, base_filters=64):
        super().__init__()

        def sn_conv(in_c, out_c, k=4, s=2, p=1):
            return nn.utils.spectral_norm(nn.Conv2d(in_c, out_c, k, s, p, bias=True))

        self.conv1 = sn_conv(1, base_filters)
        self.conv2 = sn_conv(base_filters, base_filters * 2)
        self.conv3 = sn_conv(base_filters * 2, base_filters * 4)
        self.attn = SelfAttention(base_filters * 4)
        self.conv4 = sn_conv(base_filters * 4, base_filters * 8)
        self.conv5 = sn_conv(base_filters * 8, base_filters * 16)

        self.cond_net = nn.Sequential(
            nn.utils.spectral_norm(nn.Linear(condition_dim, base_filters * 16)),
            nn.LeakyReLU(0.2, True)
        )

        self.classifier = nn.Sequential(
            nn.utils.spectral_norm(nn.Conv2d(base_filters * 32, base_filters * 16, 3, 1, 1)),
            nn.LeakyReLU(0.2, True),
            nn.AdaptiveAvgPool2d(1),
            nn.utils.spectral_norm(nn.Conv2d(base_filters * 16, 1, 1))
        )
        self.act = nn.LeakyReLU(0.2, True)

    def forward(self, img, cond):
        B = img.size(0)
        x = self.act(self.conv1(img))
        x = self.act(self.conv2(x))
        x = self.act(self.conv3(x))
        x = self.attn(x)
        x = self.act(self.conv4(x))
        x = self.act(self.conv5(x))

        c = self.cond_net(cond).view(B, -1, 1, 1).expand(-1, -1, x.size(2), x.size(3))
        combined = torch.cat([x, c], dim=1)
        return self.classifier(combined).view(B, -1)

generator = RefactoredGenerator().to(DEVICE)
discriminator = RefactoredDiscriminator().to(DEVICE)
print("Architectures compiled successfully.")

class SyntheticWheelDataset(Dataset):
    """Generates wheel image-condition pairs on the fly for GAN training."""
    def __init__(self, num_samples=1000, img_size=128):
        self.num_samples = num_samples
        self.img_size = img_size
        self.engine = UltimateWheelGenerator()
        self.patterns = ['standard', 'web', 'spiral', 'sport', 'luxury', 'hybrid']

    def __len__(self):
        return self.num_samples

    def __getitem__(self, idx):
        vol_frac = np.random.uniform(0.18, 0.40)
        num_spokes = np.random.choice([4, 5, 6, 8, 10, 12])
        complexity = np.random.uniform(0.2, 0.9)
        symmetry = np.random.uniform(0.5, 1.0)
        pattern = np.random.choice(self.patterns)

        img = self.engine.generate_wheel(
            vol_frac=vol_frac, num_spokes=num_spokes,
            size=self.img_size, pattern_type=pattern,
            complexity=complexity, symmetry=symmetry
        )
        img_tensor = torch.tensor(img, dtype=torch.float32).unsqueeze(0)
        cond_tensor = torch.tensor([vol_frac, num_spokes / 12.0, complexity, symmetry], dtype=torch.float32)
        return img_tensor, cond_tensor

dataset = SyntheticWheelDataset(num_samples=256, img_size=128)
dataloader = DataLoader(dataset, batch_size=16, shuffle=True)
print(f"Dataset ready: {len(dataset)} samples loaded into DataLoader.")

def compute_gradient_penalty(D, real_samples, fake_samples, conditions):
    alpha = torch.rand((real_samples.size(0), 1, 1, 1), device=DEVICE)
    interpolates = (alpha * real_samples + ((1 - alpha) * fake_samples)).requires_grad_(True)
    d_interpolates = D(interpolates, conditions)
    fake = torch.ones((real_samples.size(0), 1), device=DEVICE, requires_grad=False)

    gradients = torch.autograd.grad(
        outputs=d_interpolates,
        inputs=interpolates,
        grad_outputs=fake,
        create_graph=True,
        retain_graph=True,
        only_inputs=True,
    )[0]
    gradients = gradients.view(gradients.size(0), -1)
    gradient_penalty = ((gradients.norm(2, dim=1) - 1) ** 2).mean()
    return gradient_penalty

def train_wgan_gp(G, D, dataloader, epochs=2, lambda_gp=10.0):
    optimizer_G = optim.Adam(G.parameters(), lr=1e-4, betas=(0.0, 0.9))
    optimizer_D = optim.Adam(D.parameters(), lr=1e-4, betas=(0.0, 0.9))

    print("Starting WGAN-GP Training Smoke Run...")
    for epoch in range(epochs):
        for i, (real_imgs, conds) in enumerate(dataloader):
            real_imgs, conds = real_imgs.to(DEVICE), conds.to(DEVICE)
            batch_size = real_imgs.size(0)

            # Train Discriminator
            optimizer_D.zero_grad()
            z = torch.randn(batch_size, 256, device=DEVICE)
            fake_imgs = G(z, conds)

            real_validity = D(real_imgs, conds)
            fake_validity = D(fake_imgs.detach(), conds)
            gp = compute_gradient_penalty(D, real_imgs.data, fake_imgs.data, conds)

            d_loss = -torch.mean(real_validity) + torch.mean(fake_validity) + lambda_gp * gp
            d_loss.backward()
            optimizer_D.step()

            # Train Generator every 2 steps
            if i % 2 == 0:
                optimizer_G.zero_grad()
                gen_imgs = G(z, conds)
                g_loss = -torch.mean(D(gen_imgs, conds))
                g_loss.backward()
                optimizer_G.step()

        print(f"Epoch [{epoch+1}/{epochs}] | D Loss: {d_loss.item():.4f} | G Loss: {g_loss.item():.4f}")

train_wgan_gp(generator, discriminator, dataloader, epochs=2)

complexities = [0.2, 0.5, 0.8]
symmetries = [0.6, 0.8, 1.0]
fig, axes = plt.subplots(len(complexities), len(symmetries), figsize=(12, 12))

for i, comp in enumerate(complexities):
    for j, sym in enumerate(symmetries):
        wheel = wheel_engine.generate_wheel(vol_frac=0.25, num_spokes=6, complexity=comp, symmetry=sym)
        m = analyzer.comprehensive_analysis(wheel)

        axes[i, j].imshow(wheel, cmap='viridis')
        axes[i, j].set_title(f"Comp: {comp} | Sym: {sym}\nIoU Sym: {m['symmetry_avg_iou']:.3f}", fontsize=9)
        axes[i, j].axis('off')

plt.suptitle('Parameter Sweep Analysis (IoU Symmetry & Topological Integrity)', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
plt.close('all')
```

```python
# ==============================================================================
# ONE-CELL PATCH
#   - symmetry metrics -> foreground IoU (fixes inflated scores + mislabelled d1)
#   - volume adjustment -> can add AND remove, no noise speckle
#   - final touches -> no noise, strip tiny components
#   - discriminator -> real spectral normalization
# Drop after the class definitions, then rebuild ultimate_D.
# ==============================================================================

import numpy as np
from scipy import ndimage
import torch
import torch.nn as nn
from torch.nn.utils import spectral_norm

# ---------- 1. Symmetry via foreground IoU ----------
def _iou(a, b):
    inter = np.logical_and(a, b).sum()
    union = np.logical_or(a, b).sum()
    return float(inter) / float(union) if union > 0 else 0.0

def _sym_metrics(img_binary):
    h      = _iou(img_binary, np.fliplr(img_binary))
    v      = _iou(img_binary, np.flipud(img_binary))
    rot180 = _iou(img_binary, np.rot90(img_binary, 2))   # what the old "d1" actually was
    diag   = _iou(img_binary, np.rot90(img_binary, 1))   # real diagonal
    return {
        'symmetry_h': h,
        'symmetry_v': v,
        'symmetry_rot180': rot180,
        'symmetry_diag': diag,
        'symmetry_avg': (h + v + rot180 + diag) / 4,
    }

_orig_comprehensive = RefactoredDesignAnalyzer.comprehensive_analysis

def _patched_comprehensive(self, img):
    metrics = _orig_comprehensive(self, img)
    if np.sum(img > 0.5) == 0:
        return metrics
    img_binary = (img > 0.5).astype(np.uint8)
    metrics.update(_sym_metrics(img_binary))
    metrics.pop('symmetry_d1_iou', None)   # old mislabelled key
    return metrics

RefactoredDesignAnalyzer.comprehensive_analysis = _patched_comprehensive


# ---------- 2. Volume-fraction adjustment: add AND remove, no noise ----------
def _vol_adjust_v2(self, img, target_vf):
    img = (img > 0.5).astype(np.float32)
    size = img.size
    current = img.mean()
    if abs(current - target_vf) < 0.003:
        return img

    target_px = int(round(target_vf * size))
    current_px = int(img.sum())

    if current_px < target_px:
        # add material closest to existing foreground
        dist = ndimage.distance_transform_edt(1 - img)
        cand_idx = np.argwhere(img == 0)
        if len(cand_idx) == 0:
            return img
        d = dist[img == 0]
        order = np.argsort(d)
        need = min(target_px - current_px, len(order))
        pick = cand_idx[order[:need]]
        img[pick[:, 0], pick[:, 1]] = 1.0
    else:
        # remove small components first, then erode from boundary
        excess = current_px - target_px
        labeled, n = ndimage.label(img)
        if n > 0:
            sizes = np.array(ndimage.sum(img, labeled, range(1, n + 1)))
            for comp in np.argsort(sizes):          # smallest first
                if excess <= 0:
                    break
                comp_mask = labeled == (comp + 1)
                idx = np.argwhere(comp_mask)
                take = min(len(idx), excess)
                img[idx[:take, 0], idx[:take, 1]] = 0.0
                excess -= take
        while excess > 0:
            eroded = ndimage.binary_erosion(img > 0.5)
            if not eroded.any():
                break
            removed = np.argwhere((img > 0.5) & (~eroded))
            take = min(removed.shape[0], excess)
            img[removed[:take, 0], removed[:take, 1]] = 0.0
            excess -= take

    return img.astype(np.float32)


def _final_touches_v2(self, img):
    # light smoothing, re-threshold, remove specks — no random noise
    img = ndimage.gaussian_filter(img.astype(np.float32), sigma=0.4)
    img = (img > 0.5).astype(np.float32)
    labeled, n = ndimage.label(img)
    if n > 0:
        sizes = np.array(ndimage.sum(img, labeled, range(1, n + 1)))
        keep = np.where(sizes >= 4)[0] + 1
        img = np.isin(labeled, keep).astype(np.float32)
    return img


UltimateWheelGenerator._advanced_volume_adjustment = _vol_adjust_v2
UltimateWheelGenerator._apply_final_touches       = _final_touches_v2


# ---------- 3. Spectral norm on the discriminator ----------
def _sn_downsample(self, in_ch, out_ch, batch_norm=True):
    layers = [spectral_norm(nn.Conv2d(in_ch, out_ch, 4, 2, 1, bias=False))]
    if batch_norm:
        layers.append(nn.BatchNorm2d(out_ch))
    layers.append(nn.LeakyReLU(0.2, True))
    return nn.Sequential(*layers)

if 'UltimateDiscriminator' in globals():
    UltimateDiscriminator._downsample_block = _sn_downsample


# ---------- 4. Sanity check ----------
print("Patched:")
print("  symmetry metrics -> foreground IoU (symmetry_rot180, symmetry_diag)")
print("  volume adjustment -> adds AND removes, no speckle")
print("  final touches -> no noise, specks stripped")
print("  discriminator -> real spectral normalization")
print("\nRebuild the discriminator for spectral norm to take effect:")
print("  discriminator = RefactoredDiscriminator().to(DEVICE)\n")

# smoke test
w = wheel_engine.generate_wheel(
    vol_frac=0.25, num_spokes=8, pattern_type='standard',
    complexity=0.5, symmetry=0.8,
)
m = analyzer.comprehensive_analysis(w)
print(f"wheel: vf={m['volume_fraction']:.4f}  components={m['num_components']}  "
      f"sym_avg={m['symmetry_avg']:.4f}  sym_rot180={m['symmetry_rot180']:.4f}")
```

```python
import torch
import numpy as np
import matplotlib.pyplot as plt

# Ensure generator is in evaluation mode
generator.eval()

# Define the parameter sweep values
volume_fractions = [0.20, 0.28, 0.35]
spoke_counts = [4, 8, 12]
fixed_complexity = 0.6
fixed_symmetry = 0.9

fig, axes = plt.subplots(len(volume_fractions), len(spoke_counts), figsize=(12, 12))

# Generate and plot wheels for each combination
with torch.no_grad():
    for r_idx, vf in enumerate(volume_fractions):
        for c_idx, spokes in enumerate(spoke_counts):
            # Construct the 4-dimensional condition tensor: [vol_frac, normalized_spokes, complexity, symmetry]
            cond = torch.tensor([[vf, spokes / 12.0, fixed_complexity, fixed_symmetry]], dtype=torch.float32, device=DEVICE)

            # Use a fixed noise vector per column to isolate the effect of condition variations
            torch.manual_seed(42 + c_idx)
            z = torch.randn(1, 256, device=DEVICE)

            # Synthesize
            gen_img = generator(z, cond).cpu().squeeze().numpy()

            # Plotting
            ax = axes[r_idx, c_idx]
            ax.imshow(gen_img, cmap='gray')
            ax.set_title(f"VF: {vf:.2f} | Spokes: {spokes}\nComp: {fixed_complexity} | Sym: {fixed_symmetry}", fontsize=10)
            ax.axis('off')

plt.suptitle("Generator Sweep Grid: Volume Fraction (Rows) vs. Spoke Counts (Columns)", fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```
